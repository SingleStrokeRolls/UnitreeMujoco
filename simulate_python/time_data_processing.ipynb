```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('./time_data.csv')
df.head()
```

```python
print(df['predictino_time'].mean())
```

```python
x = np.arange(len(df))  # 0, 1, 2, ..., 199

plt.figure(figsize=(10, 6))

# 绘制两列数据
plt.scatter(x, df['duration'], label='Duration', alpha=0.6, s=20, c='blue')
plt.scatter(x, df['predictino_time'], label='Prediction Time', alpha=0.6, s=20, c='red')

plt.xlabel('Iterations', fontsize=12)
plt.ylabel('Time(s)', fontsize=12)
plt.legend()
plt.grid(True, linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
```

```python
condition = df['predictino_time'] < df['duration'] * 0.6
count = condition.sum()  # 或者 len(df[condition])
count
```

```python
df['percentage'] = df['predictino_time'] / df['duration']
plt.hist(df['percentage'], bins=20, edgecolor='black')
plt.xlabel('Prediction Response Time / Duration (%)')
plt.ylabel('Counts')
plt.show()
```
